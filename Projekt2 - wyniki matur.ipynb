<h1>Podsumowanie powyższych danych</h1>
<p style="font-size: 20px;">Najwięcej szkół, w których uczniowie zdawali maturę było w województwie mazowieckim. Miejscowości z największą liczbą szkół to kolejeno Warszawa, Kraków, Poznań, Łódź, Wrocław. Jeśli chodzi o typ placówki to najwięcej było liceów.</p>

```python
def pass_percentage(df):
    total = (df['Liczba zdających'] * df['Procent zdawalności']).sum(skipna=True)
    number_of_candidates = df['Liczba zdających'].sum(skipna=True)
    return total / number_of_candidates
```

```python
def plot_pass_percentage():
    years = [2021, 2022, 2023, 2024]
    pass_percentages = [
        pass_percentage(df_2021),
        pass_percentage(df_2022),
        pass_percentage(df_2023),
        pass_percentage(df_2024)
    ]

    colors = ['skyblue', 'lightgreen', 'lightcoral', 'gold']

    plt.figure(figsize=(8, 6))
    plt.bar(years, pass_percentages, color=colors)

    plt.title('Procent zdawalności matury w Latach 2021-2024', fontsize=14)
    plt.xlabel('Rok', fontsize=12)
    plt.ylabel('Procent zdawalności', fontsize=12)

    plt.xticks(years)

    for i, v in enumerate(pass_percentages):
        plt.text(years[i], v + 1, f'{v:.2f}%', ha='center', fontsize=10)

    plt.show()
```

<h1>Wykres przedstawiający procent zdawalności matury w latach 2021-2024</h1>

```python
plot_pass_percentage()
```

<p style="font-size: 20px;">Widzimy, że najgorsza zdawalność matur była w latach 2021 i 2022, a najlepsza w 2023 i 2024.</p>

```python
def pass_percentage_by_region(df):
    unique_regions = df['Województwo'].unique()
    results = {}
    
    for region in unique_regions:
        regional_data = df[df['Województwo'] == region]
        pass_percentage = (regional_data['Liczba zdających'] * regional_data['Procent zdawalności']).sum() / regional_data['Liczba zdających'].sum()
        results[region] = pass_percentage
    
    return pd.Series(results)
```

```python
def plot_pass_percentage_region(df):

    result = pass_percentage_by_region(df)

    plt.figure(figsize=(10, 6)) 
    result.plot(kind='bar', color='lightgreen', edgecolor='black')

    plt.title('Procent zdawalności w poszczególnych województwach', fontsize=14)
    plt.xlabel('Województwo', fontsize=12)
    plt.ylabel('Procent zdawalności', fontsize=12)

    for index, value in enumerate(result):
        plt.text(index, value + 1, f'{value:.2f}%', ha='center', fontsize=10)

    plt.xticks(rotation=45, ha='right') 
    plt.tight_layout()
    plt.show()
```

<h1>Wykres przedstawiający procent zdawalności w poszczególnych województwach w 2021 roku</h1>

<p style="font-size: 20px;">Najwyższa zdawalność jest w województwie mazowieckim i małopolskim.</p>

```python
def pass_percentage_in_compulsory_subjects(df, text):
    total = (df['Liczba zdających'] * df[text]).sum(skipna=True)
    number_of_candidates = df['Liczba zdających'].sum(skipna=True)
    return total / number_of_candidates
```

```python
def plot_pass_percentage_in_compulsory_subjects(df, year):
    pass_percentages_in_compulsory_subjects = [
        pass_percentage_in_compulsory_subjects(df, "Procent zdawalności języka polskiego podstawowego"),
        pass_percentage_in_compulsory_subjects(df, "Procent zdawalności z języka angielskiego podstawowego"),
        pass_percentage_in_compulsory_subjects(df, "Procent zdawalności z matematyki podstawowej")
    ]
    subjects = ['Język polski', 'Język angielski', 'Matematyka']
    colors = ['lightgreen', 'lightcoral', 'gold']

    plt.figure(figsize=(8, 6))
    plt.bar(subjects, pass_percentages_in_compulsory_subjects, color=colors)

    plt.title(f'Procent zdawalności przedmiotów obowiązkowych w {year} roku', fontsize=14)
    plt.xlabel('Przedmiot', fontsize=12)
    plt.ylabel('Procent zdawalności', fontsize=12)

    plt.xticks(subjects)

    for i, v in enumerate(pass_percentages_in_compulsory_subjects):
        plt.text(subjects[i], v + 1, f'{v:.2f}%', ha='center', fontsize=10)

    plt.show()
```

```python
def pass_percentage_100_regions(df):
    filtered_df = df[(df["Procent zdawalności"] == 100)]
    return filtered_df.groupby("Województwo").size()
```

```python
def plot_pass_percentage_100_regions(df, year):
    data = pass_percentage_100_regions(df)
    ax = data.plot(
        kind="bar", 
        figsize=(10, 6), 
        color="gold", 
        title=f'Liczba szkół ze 100% zdawalnością w {year} roku w poszczególnych województwach'
    )

    for i, value in enumerate(data):
        plt.text(i, value + 0.5, str(value), ha='center', fontsize=10)

    plt.xlabel("Województwo")
    plt.ylabel("Liczba szkół")
    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()
```

Najmniej zaś jest z województwa opolskiego, które jest najmniejsze w Polsce pod względem powierzhcni oraz ludności. Dla większości województw wraz z kolejnymi latami jest więcej szkół ze 100% zdawalnością.</p>

```python
def avarage_score_in_extended_subjects(df, column_name):
    df_cleaned = df.dropna(subset=['Liczba zdających', column_name])
    total = (df_cleaned['Liczba zdających'] * df_cleaned[column_name]).sum()
    number_of_candidates = df_cleaned['Liczba zdających'].sum()

    return total / number_of_candidates
```

```python
def plot_pass_percentage_for_subjects(df, subjects, column_names, year):
    results = []
    
    for column_name in column_names:
        avarage_score = avarage_score_in_extended_subjects(df, column_name)
        results.append(avarage_score)

    plt.figure(figsize=(12, 6))
    plt.bar(subjects, results, color='lightgreen')
    plt.title(f'Średnie wyniki z przedmiotów rozszerzonych w {year} roku')
    plt.xlabel('Przedmioty')
    plt.ylabel('Średni wynik')
    plt.xticks(rotation=45, ha='center')
    
    for i, value in enumerate(results):
        plt.text(i, value + 0.5, f'{value:.2f}%', ha='center', fontsize=9)
    
    plt.tight_layout()
    plt.show()
```

```python
column_names = ["Średni wynik z biologii", "Średni wynik z chemii", "Średni wynik z filozofii", "Średni wynik z fizyki", "Średni wynik z geografii", "Średni wynik z historii", "Średni wynik z historii muzyki", "Średni wynik z historii sztuki", "Średni wynik z informatyki", "Średni wynik z języka angielskiego rozszerzonego", "Średni wynik z języka polskiego rozszerzonego", "Średni wynik z matematyki rozszerzonej", "Średni wynik z WOS-u" ]
subjects = ["Biologia", "Chemia", "Filozofia", "Fizyka", "Geografia", "Historia", "Historia muzyki", "Historia sztuki", "Informatyka", "Język angielski", "Język polski", "Matematyka", "Wos"]
```

<h1>Wykresy przedstawiające średnie wyniki z przedmiotów rozszerzonych w kolejnych latach</h1>

```python
plot_pass_percentage_for_subjects(df_2021, subjects, column_names, 2021)
```

```python
plot_pass_percentage_for_subjects(df_2022, subjects, column_names, 2022)
```
